```python
import pandas as pd
import wbdata
import datetime
```

Script to get reqired data from World Bank 

```python
indicators = {
    "SP.POP.TOTL": "population",
    "NY.GDP.MKTP.KD": "gdp_constant_usd",
    "NY.GDP.PCAP.KD": "gdp_per_capita",
    "FP.CPI.TOTL.ZG": "inflation_rate",
    "NE.RSB.GNFS.CD": "trade_balance",
    "DT.DOD.DECT.CD": "external_debt",
    "SI.POV.GINI": "income_inequality_gini"
}
date_time = (datetime.datetime(2000, 1, 1), datetime.datetime(2021, 12, 31))
try:
    df = wbdata.get_dataframe(indicators, date=date_time)
except Exception as e:
    print(f"Error fetching data: {e}")
    # Try fetching without date parameter as fallback
    df = wbdata.get_dataframe(indicators)
df.reset_index(inplace = True)
df.columns
```
